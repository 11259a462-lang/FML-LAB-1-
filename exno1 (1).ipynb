```python

import numpy as np

# 1. One-dimensional array
print("1D Array:")
print(np.array([1, 2, 3, 4, 5]))

# 2. Two-dimensional array
print("\n2D Array:")
print(np.array([[1, 2, 3], [4, 5, 6]]))

# 3. Array of zeros
print("\nZeros:")
print(np.zeros((3, 3), dtype=int))

# 4. Array of ones
print("\nOnes:")
print(np.ones((2, 4), dtype=int))

# 5. Identity matrix
print("\nIdentity Matrix:")
print(np.eye(3, dtype=int))

# 6. Range of numbers
print("\nRange:")
print(np.arange(0, 10, 2))
```

```python
import numpy as np

# Create two arrays
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Display arrays and perform operations
print("Array a:", a)
print("Array b:", b)

print("\nAddition (a + b):", a + b)
print("Multiplication (a * b):", a * b)
print("Scalar Multiplication (a * 10):", a * 10)

print("\nSquare Root of a:", np.sqrt(a))
print("Mean of a:", np.mean(a))
print("Dot Product (a . b):", np.dot(a, b))
```

```python
import numpy as np

# Create a 3x3 matrix
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# Display the matrix
print("Matrix:\n", matrix)

# Access an element
print("\nElement at [1, 2]:", matrix[1, 2])

# Display the first row
print("\nFirst Row:", matrix[0, :])

# Display the second column
print("\nSecond Column:", matrix[:, 1])

# Display the top-left 2x2 sub-matrix
print("\nTop-left 2x2 Sub-matrix:\n", matrix[0:2, 0:2])
```
