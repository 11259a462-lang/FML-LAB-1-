```python
import pandas as pd

# Create a dictionary
data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "City": ["New York", "Paris", "London"]
}

# Create a DataFrame
df = pd.DataFrame(data)

# Display the first five rows
print("1. First Five Rows:")
print(df.head().to_string(index=False))

# Display information about the DataFrame
print("\n2. DataFrame Information:")
df.info()

# Display statistical summary
print("\n3. Statistical Summary:")
print(df.describe())
```

```python

import pandas as pd

# Create a DataFrame
data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "City": ["New York", "Paris", "London"]
}

df = pd.DataFrame(data)

# Select the Age column
ages = df["Age"]
print("Ages column:")
print(ages)

print("\n" + "-" * 30)

# Filter rows where Age > 25
above_25 = df[df["Age"] > 25]
print("People with Age > 25:")
print(above_25)

print("\n" + "-" * 30)

# iloc: Select the first row by integer position
row_0 = df.iloc[0]
print("First row using iloc (row index 0):")
print(row_0)

print("\n" + "-" * 30)

# loc: Select the name at index label 0
specific_val = df.loc[0, "Name"]
print("Name of the person at index 0:")
print(specific_val)
```

```python

import pandas as pd
import numpy as np

# 1. Create a sample dataset with missing values
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [23, np.nan, 22, 28, np.nan],
    "City": ["New York", "London", np.nan, "Paris", "Berlin"]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
print("\n" + "-" * 50 + "\n")

# 2. Check for null values
print("Null values count in each column:")
print(df.isnull().sum())
print("\n" + "-" * 50 + "\n")

# 3. Drop rows with missing values
df_clean = df.dropna()

print("DataFrame after dropping rows with missing values:")
print(df_clean)
print("\n" + "-" * 50 + "\n")

# 4. Fill missing values with 0
df_filled = df.fillna(0)

print("DataFrame after filling missing values with 0:")
print(df_filled)
print("\n" + "-" * 50 + "\n")

# 5. Fill missing Age values with the mean
df_mean_filled = df.copy()
df_mean_filled["Age"] = df_mean_filled["Age"].fillna(
    df_mean_filled["Age"].mean()
)

print("DataFrame after filling Age with mean value:")
print(df_mean_filled)
print("\n" + "-" * 50 + "\n")

# 6. Fill missing City values with "Unknown"
df_city_filled = df.copy()
df_city_filled["City"] = df_city_filled["City"].fillna("Unknown")

print("DataFrame after filling City with 'Unknown':")
print(df_city_filled)
```
