```python

# STEP 1: IMPORT LIBRARIES
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# STEP 2: CREATE DATASET
data = {
    'Student': ['S1', 'S2', 'S3', 'S4', 'S5',
                'S6', 'S7', 'S8', 'S9', 'S10'],
    'Classes_Attended': [30, 35, 40, 45, 50,
                         55, 60, 65, 70, 75],
    'Internal_Marks': [35, 38, 42, 46, 50,
                       55, 60, 65, 70, 75]
}

df = pd.DataFrame(data)

print("Student Dataset:")
print(df)

# STEP 3: FEATURES AND TARGET
X = df[['Classes_Attended']]
y = df['Internal_Marks']

# STEP 4: TRAIN-TEST SPLIT
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=7
)

# STEP 5: TRAIN MODEL
model = LinearRegression()
model.fit(X_train, y_train)

# STEP 6: MAKE PREDICTIONS
predictions = model.predict(X_test)

print("\nActual Marks:", list(y_test))
print("Predicted Marks:", predictions.round(2))

# STEP 7: EVALUATE MODEL
mse = mean_squared_error(y_test, predictions)

print("\nMean Squared Error (MSE):", round(mse, 2))
print("Marks per Class:", round(model.coef_[0], 4))
print("Base Marks:", round(model.intercept_, 4))

# STEP 8: VISUALIZATION
plt.figure(figsize=(8, 5))

plt.scatter(
    X['Classes_Attended'],
    y,
    color='blue',
    label='Actual Data'
)

plt.plot(
    X['Classes_Attended'],
    model.predict(X),
    color='red',
    label='Regression Line'
)

plt.xlabel("Classes Attended")
plt.ylabel("Internal Marks")
plt.title("Classes Attended vs Internal Marks")
plt.legend()
plt.grid(True)
plt.show()
```
