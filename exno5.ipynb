```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# STEP 1: LOAD DATASET
data = pd.read_csv("data.csv")

# STEP 2: DISPLAY FIRST 5 ROWS
print("Original Dataset:")
print(data.head())

# STEP 3: CHECK MISSING VALUES
print("\nMissing Values:")
print(data.isnull().sum())

# STEP 4: FILL MISSING NUMERICAL VALUES WITH THE MEAN
numeric_columns = data.select_dtypes(include="number").columns

data[numeric_columns] = data[numeric_columns].fillna(
    data[numeric_columns].mean()
)

# STEP 5: SELECT REQUIRED NUMERICAL COLUMNS
required_columns = [
    "Hours_Studied",
    "Attendance",
    "Assignments",
    "Marks"
]

X = data[required_columns].copy()

# STEP 6: NORMALIZE DATA
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# STEP 7: CONVERT BACK TO DATAFRAME
X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns,
    index=X.index
)

# STEP 8: DISPLAY NORMALIZED DATA
print("\nNormalized Dataset:")
print(X_scaled.head())
```
